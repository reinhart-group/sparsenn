```python
import torch
from sklearn import model_selection, preprocessing

# aux_targets = ['vdw_radius_uff','Heat_Capacity', 'normUSFE']
aux_targets = ['vdw_radius_mm3', 'en_allen', 'No_Spectral_lines']  # goldschmidt technically requires a classifier...
# aux_targets = []
y_a = xf.loc[:, aux_targets]

tasks = ['tensile ductility (%)']
# tasks = ['log fracture toughness (Pa m^0.5)']
# tasks = ['log fracture toughness (Pa m^0.5)', ]

aux_features = ['vdw_radius_uff', 'atomic_weight', 'Ion_Pot_1']
# aux_features = ['series_id', 'block']
x_a = xf.loc[:, aux_features]

# x = imp_df.loc[:, ['Temperature (K)'] + phase_list + list(ele_list)]
# x[np.isnan(x)] = 0  # fill missing elemental composition with 0's

x = pd.concat([imp_df.loc[:, ['Temperature (K)'] + phase_list + list(ele_list)], x_a], axis=1)
x[np.isnan(x)] = 0  # fill missing elemental composition with 0's

# x = xf  # use the rule of mixtures feature set

y = pd.concat([imp_df.loc[:, tasks], y_a], axis=1)

rs = 0
x_train, x_vt, y_train, y_vt = model_selection.train_test_split(x, y, train_size=0.6, random_state=rs)
x_valid, x_test, y_valid, y_test = model_selection.train_test_split(x_vt, y_vt, train_size=0.5, random_state=rs)

x_scaler = preprocessing.StandardScaler().fit(x_train.values)
y_scaler = preprocessing.StandardScaler().fit(y_train.values)

x_train = torch.tensor(x_scaler.transform(x_train.values)).float()
y_train = torch.tensor(y_scaler.transform(y_train.values)).float()

x_valid = torch.tensor(x_scaler.transform(x_valid.values)).float()
y_valid = torch.tensor(y_scaler.transform(y_valid.values)).float()

x_test = torch.tensor(x_scaler.transform(x_test.values)).float()
y_test = torch.tensor(y_scaler.transform(y_test.values)).float()

print(x_train.shape, x_valid.shape, x_test.shape)
print(y_train.shape, y_valid.shape, y_test.shape)
```

```python
from torch import nn, optim
import sparsenn
from sparsenn import pareto
from copy import deepcopy

input_dim = x_train.shape[1]
hidden_dim = (32, 16, 8, )
output_dim = y_train.shape[1]

model = sparsenn.RegularizedMLPwithSkips(input_dim, hidden_dim, output_dim, activation=nn.ReLU())
criterion = sparsenn.MultiTaskMSELoss()
optimizer = optim.Adam(model.parameters(), lr=1e-2)

explored_values = []
frontier = []

max_epochs = int(1e5)

best_model = deepcopy(model)

main_task_id = 0

lmbda_reg = 1e-5
reg_prec = 3
nwrite = 1000
patience_interval = nwrite * 5
last_best = 0
last_valid_best = 0
best_valid = np.inf
model.train()
for epoch in range(max_epochs):
    model.train()  # un-clamp the weights
    
    optimizer.zero_grad()
    outputs = model(x_train)
    loss = criterion(outputs, y_train)
    reg_loss = model.regularization_loss()
    # if epoch < 1e3:
    #     reg_loss *= 0.

    total_loss = loss + lmbda_reg * reg_loss
    total_loss.backward()
    optimizer.step()
    
    model.eval()  # clamp the weights
    with torch.no_grad():
        outputs = model(x_valid)
        valid_loss = criterion(outputs[:, main_task_id], y_valid[:, main_task_id])
    
    point = np.array([valid_loss.item(), np.round(reg_loss.item(), reg_prec)])
    explored_values.append(point)
    if not pareto.is_dominated(point, frontier):
        frontier = pareto.update_pareto_front(point, frontier)
        last_best = epoch

        if valid_loss < best_valid:
            best_valid = valid_loss
            last_valid_best = epoch
            best_model = deepcopy(model)

    if reg_loss.item() < 1:
        break

    if (epoch+1) % nwrite == 0:
        with torch.no_grad():
            outputs = model(x_test)
            test_loss = criterion(outputs[:, main_task_id], y_test[:, main_task_id])
        print(f'Epoch {epoch+1}, MSE: {loss.item():.4f}, L0R: {reg_loss.item():.2f}, ValMSE: {valid_loss.item():.4f}, TestMSE: {test_loss.item():.4f}')
    
    if epoch - last_best > patience_interval:
        print('Patience expired...')
        break
    
    if epoch - last_valid_best > 10 * patience_interval:
        print('Meta-Patience expired...')
        break
```

```python
import torch
import torch.optim as optim

class PCGrad:
    def __init__(self, model, num_tasks, optimizer, criterion, lambda_reg=0.):
        self.model = model
        self.num_tasks = num_tasks
        self.base_optimizer = optimizer
        self.criterion = criterion
        self.lambda_reg = lambda_reg
    
    def compute_losses(self, output, target):
        if self.lambda_reg > 0:
            reg_loss = self.model.regularization_loss()
        else:
            reg_loss = 0.
        losses = []
        for i in range(self.num_tasks):
            task_output = output[:, i].unsqueeze(1)
            task_target = target[:, i].unsqueeze(1)
            task_loss = self.criterion(task_output, task_target)
            total_loss = task_loss + self.lambda_reg * reg_loss / self.num_tasks
            losses.append(total_loss)
        return losses
    
    def compute_gradients(self, losses):
        grads = []
        for loss in losses:
            self.base_optimizer.zero_grad()
            loss.backward(retain_graph=True)
            grad = []
            for param in self.model.parameters():
                if param.grad is not None:
                    grad.append(param.grad.view(-1))
            grads.append(torch.cat(grad))
        return torch.stack(grads)
    
    def project_gradients(self, grads):
        proj_grads = grads.clone()
        for i in range(self.num_tasks):
            for j in range(self.num_tasks):
                if i != j:
                    inner_product = torch.dot(proj_grads[i], grads[j])
                    proj_direction = inner_product / torch.dot(grads[j], grads[j])
                    proj_grads[i] -= torch.clamp(proj_direction, min=0) * grads[j]
        return proj_grads.mean(dim=0)
    
    def apply_gradients(self, grad):
        idx = 0
        for param in self.model.parameters():
            if param.grad is not None:
                size = param.grad.numel()
                param.grad.data = grad[idx:idx+size].view_as(param.data)
                idx += size
    
    def train_step(self, inputs, targets):
        self.model.train()
        outputs = self.model(inputs)
        losses = self.compute_losses(outputs, targets)
        grads = self.compute_gradients(losses)
        pcgrad = self.project_gradients(grads)
        self.apply_gradients(pcgrad)
        self.base_optimizer.step()
        return [loss.item() for loss in losses]


class PCGradWithRegTask(PCGrad):
    
    def compute_losses(self, output, target):
        if self.lambda_reg > 0:
            reg_loss = self.model.regularization_loss()
        else:
            reg_loss = 0.
        losses = []
        for i in range(self.num_tasks):
            task_output = output[:, i].unsqueeze(1)
            task_target = target[:, i].unsqueeze(1)
            task_loss = self.criterion(task_output, task_target)
            # total_loss = task_loss + self.lambda_reg * reg_loss / self.num_tasks
            # losses.append(total_loss)
            losses.append(task_loss)
        losses.append(reg_loss)
        return losses
    
    def compute_gradients(self, losses):
        grads = []
        for loss in losses:
            self.base_optimizer.zero_grad()
            loss.backward(retain_graph=True)
            grad = []
            for param in self.model.parameters():
                if param.grad is not None:
                    grad.append(param.grad.view(-1))
                else:
                    grad.append(torch.zeros_like(param.view(-1)))
            grads.append(torch.cat(grad))
        return torch.stack(grads)    
    
    
class wPCGrad(PCGrad):
    def __init__(self, model, num_tasks, optimizer, criterion, lambda_reg=0., task_weights=None):
        super().__init__(model, num_tasks, optimizer, criterion, lambda_reg)
        if task_weights is None:
            self.task_weights = torch.ones(num_tasks) / num_tasks
        else:
            self.task_weights = task_weights 
    
    def update_task_weights(self, new_weights):
        self.task_weights = new_weights
    
    def project_gradients(self, grads):
        proj_grads = grads.clone()
        
        # Sample a task to remain unmodified based on task weights
        unmodified_task = torch.multinomial(self.task_weights, 1).item()
        
        for i in range(self.num_tasks):
            if i != unmodified_task:
                inner_product = torch.dot(proj_grads[i], grads[unmodified_task])
                if inner_product < 0:
                    proj_direction = inner_product / torch.dot(grads[unmodified_task], grads[unmodified_task])
                    proj_grads[i] -= proj_direction * grads[unmodified_task]
        
        return proj_grads.mean(dim=0)
    
    def train_step(self, inputs, targets):
        self.model.train()
        outputs = self.model(inputs)
        losses = self.compute_losses(outputs, targets)
        grads = self.compute_gradients(losses)
        wpcgrad = self.project_gradients(grads)
        self.apply_gradients(wpcgrad)
        self.base_optimizer.step()
        return [loss.item() for loss in losses]
    
class DynamicTaskPrioritization:
    def __init__(self, num_tasks, gamma=2, epsilon=1e-8):
        self.num_tasks = num_tasks
        self.gamma = gamma
        self.epsilon = epsilon
        self.task_losses = np.zeros(num_tasks)
        self.task_weights = np.ones(num_tasks) / num_tasks

    def update_task_losses(self, new_losses):
        """
        Update the task losses with the latest values.
        :param new_losses: List or numpy array of new loss values for each task
        """
        self.task_losses = np.array(new_losses)

    def compute_weights(self):
        """
        Compute new task weights based on the current task losses.
        """
        # Avoid division by zero
        adjusted_losses = self.task_losses + self.epsilon
        
        # Compute weights according to the formula in the paper
        weights = adjusted_losses ** self.gamma
        normalized_weights = weights / np.sum(weights)
        
        self.task_weights = normalized_weights
        return self.task_weights

    def get_torch_weights(self):
        """
        Return the current weights as a PyTorch tensor.
        """
        return torch.FloatTensor(self.task_weights)    
```

```python
input_dim = x_train.shape[1]
hidden_dim = (16, 24, 32, )
output_dim = y_train.shape[1]
model = sparsenn.RegularizedMLPwithSkips(input_dim, hidden_dim, output_dim, activation=nn.ReLU())
optimizer = optim.Adam(model.parameters(), lr=1e-2)
pcgrad_mtl = wPCGrad(model, output_dim, optimizer, criterion, lambda_reg=1e-5)

dtp = DynamicTaskPrioritization(output_dim, gamma=2)
task_priority = [0]

explored_values = []
frontier = []

last_best = 0
last_valid_best = 0
best_valid = np.inf
best_model = None

# Training loop
num_epochs = int(1e5)
n_write = 100
patience_interval = 100 * n_write
for epoch in range(num_epochs):
    losses = pcgrad_mtl.train_step(x_train, y_train)
    
    dtp.update_task_losses(losses)
    new_weights = dtp.compute_weights()
    pcgrad_mtl.update_task_weights(dtp.get_torch_weights())

    m = pcgrad_mtl.model
    with torch.no_grad():
        m.eval()
        reg_loss = m.regularization_loss()
        outputs = m(x_valid)
        valid_loss = criterion(outputs[:, task_priority], y_valid[:, task_priority])
        m.train()
    
    # point = np.array([valid_loss.item(), np.round(reg_loss.item(), reg_prec)])
    point = np.array([valid_loss.item()])
    explored_values.append(point)
    if not pareto.is_dominated(point, frontier):
        frontier = pareto.update_pareto_front(point, frontier)
        last_best = epoch

        if valid_loss < best_valid:
            best_valid = valid_loss
            best_model = deepcopy(m)
            last_valid_best = epoch

    if reg_loss.item() < 1:
        break

    if epoch - last_best > patience_interval:
        print('Patience expired...')
        break

    if epoch - last_valid_best > 10 * patience_interval:
        print('Meta-Patience expired...')
        break

    if (epoch+1) % n_write == 0:
        m.eval()
        with torch.no_grad():
            loss = criterion(m(x_train)[:, task_priority], y_train[:, task_priority])
            valid_loss = criterion(m(x_valid)[:, task_priority], y_valid[:, task_priority])
            test_loss = criterion(m(x_test)[:, task_priority], y_test[:, task_priority])
            reg_loss = m.regularization_loss()
        print(f'Epoch {epoch+1}, MSE: {loss.item():.4f}, L0R: {reg_loss.item():.2f}, ValMSE: {valid_loss.item():.4f}, TestMSE: {test_loss.item():.4f}')
        m.train()
```

```python
from torch.nn import functional as F

class AWD(PCGrad):
    def __init__(self, model, num_tasks, optimizer, criterion, gamma=0.01, lambda_reg=0.):
        super().__init__(model, num_tasks, optimizer, criterion, lambda_reg)
        self.gamma = gamma
        
        # Initialize redundant vector with current model parameters
        base_params = self._get_param_vector()
        self.redundant_vector = nn.Parameter(base_params.clone())
        
        # Initialize disentangled vectors for each task
        self.disentangled_vectors = nn.ParameterList([
            nn.Parameter(torch.zeros_like(base_params))
            for _ in range(num_tasks)
        ])
        
        # Create new optimizer including model parameters
        self.base_optimizer = type(optimizer)([
            {'params': model.parameters()},  # Include model parameters
            {'params': self.redundant_vector, 'lr': optimizer.defaults['lr'] * 0.1},
            {'params': self.disentangled_vectors.parameters()}
        ], **{k:v for k,v in optimizer.defaults.items() if k != 'params'})

    def train_step(self, inputs, targets):
        self.model.train()
        self.base_optimizer.zero_grad()
        
        losses_by_task = []
        
        # Compute losses for each task
        for task_id in range(self.num_tasks):
            # Set task-specific parameters
            task_params = self.redundant_vector + self.disentangled_vectors[task_id]
            self._set_param_vector(task_params)
            
            # Forward pass
            outputs = self.model(inputs)
            task_output = outputs[:, task_id].unsqueeze(1)
            task_target = targets[:, task_id].unsqueeze(1)
            task_loss = self.criterion(task_output, task_target)
            
            # Add orthogonality loss
            ortho_loss = self.orthogonality_loss()
            total_loss = task_loss + self.gamma * ortho_loss
            
            # Accumulate gradients
            total_loss.backward()
            losses_by_task.append(task_loss.item())
        
        # Update parameters
        self.base_optimizer.step()
        self.base_optimizer.zero_grad()
        
        return losses_by_task

    def _get_param_vector(self):
        """Get flattened parameter vector from model"""
        params = []
        for param in self.model.parameters():
            params.append(param.data.view(-1))
        return torch.cat(params)

    def _set_param_vector(self, vector):
        """Set model parameters from flattened vector"""
        idx = 0
        for param in self.model.parameters():
            size = param.numel()
            param.data = vector[idx:idx+size].view_as(param.data)
            idx += size

    def orthogonality_loss(self):
        """Calculate orthogonality loss between disentangled vectors"""
        loss = 0
        for i in range(self.num_tasks):
            for j in range(i + 1, self.num_tasks):
                cos_sim = F.cosine_similarity(
                    self.disentangled_vectors[i].flatten(),
                    self.disentangled_vectors[j].flatten(),
                    dim=0
                )
                loss += torch.abs(cos_sim)
        return loss / (self.num_tasks * (self.num_tasks - 1) / 2)
    
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F

class AWD(object):
    def __init__(self, model, num_tasks, optimizer, criterion, alpha=1e-4, lr_scale=0.1):
        """
        Initialize AWD.
        
        Args:
            model: Base model to be merged
            num_tasks: Number of tasks
            optimizer: Base optimizer (e.g., Adam)
            criterion: Loss function
            alpha: Weight for redundant vector norm constraint (default: 1e-4)
            lr_scale: Learning rate scale for redundant vector (default: 0.1)
        """
        self.model = model
        self.num_tasks = num_tasks
        self.criterion = criterion
        self.alpha = alpha
        
        # Initialize redundant vector with current model parameters
        base_params = self._get_param_vector()
        self.redundant_vector = nn.Parameter(base_params.clone())
        
        # Initialize disentangled vectors for each task
        self.disentangled_vectors = nn.ParameterList([
            nn.Parameter(torch.zeros_like(base_params))
            for _ in range(num_tasks)
        ])
        
        # Create optimizer including model parameters
        self.optimizer = type(optimizer)([
            {'params': model.parameters()},
            {'params': self.redundant_vector, 'lr': optimizer.defaults['lr'] * lr_scale},
            {'params': self.disentangled_vectors.parameters()}
        ], **{k:v for k,v in optimizer.defaults.items() if k != 'params'})

    def _get_param_vector(self):
        """Get flattened parameter vector from model"""
        params = []
        for param in self.model.parameters():
            if param.requires_grad:
                params.append(param.data.view(-1))
        return torch.cat(params)

    def _set_param_vector(self, vector):
        """Set model parameters from flattened vector"""
        idx = 0
        for param in self.model.parameters():
            if param.requires_grad:
                size = param.numel()
                param.data = vector[idx:idx+size].view_as(param.data)
                idx += size

    def orthogonality_loss(self):
        """Calculate orthogonality loss between disentangled vectors"""
        loss = 0
        for i in range(self.num_tasks):
            for j in range(i + 1, self.num_tasks):
                cos_sim = F.cosine_similarity(
                    self.disentangled_vectors[i].flatten(),
                    self.disentangled_vectors[j].flatten(),
                    dim=0
                )
                loss += torch.abs(cos_sim)
        return loss / (self.num_tasks * (self.num_tasks - 1) / 2)

    def redundancy_loss(self):
        """Calculate norm constraint loss for redundant vector"""
        return torch.norm(self.redundant_vector)

    def train_step(self, inputs, targets):
        """Perform one training step"""
        self.model.train()
        self.optimizer.zero_grad()
        
        # Store original parameters
        orig_params = self._get_param_vector()
        
        total_loss = 0
        task_losses = []
        
        # Compute losses for each task
        for task_id in range(self.num_tasks):
            # Set task-specific parameters
            task_params = self.redundant_vector + self.disentangled_vectors[task_id]
            self._set_param_vector(task_params)
            
            # Forward pass
            outputs = self.model(inputs)
            task_output = outputs[:, task_id].unsqueeze(1)
            task_target = targets[:, task_id].unsqueeze(1)
            task_loss = self.criterion(task_output, task_target)
            
            task_losses.append(task_loss.item())
            total_loss += task_loss

        # Add orthogonality and redundancy losses
        ortho_loss = self.orthogonality_loss()
        red_loss = self.redundancy_loss()
        total_loss = total_loss + ortho_loss + self.alpha * red_loss
        
        # Backward pass and update
        total_loss.backward()
        self.optimizer.step()
        
        # Restore original parameters
        self._set_param_vector(orig_params)
        
        return task_losses

    def get_merged_model(self, coefficient=0.3):
        """
        Get merged model with specified coefficient.
        
        Args:
            coefficient: Scaling factor for task vectors (default: 0.3)
        """
        merged_params = self.redundant_vector.clone()
        for vec in self.disentangled_vectors:
            merged_params += coefficient * vec
        
        # Create new model with merged parameters
        merged_model = type(self.model)(
            *self.model.__init_args__,
            **self.model.__init_kwargs__
        )
        with torch.no_grad():
            self._set_param_vector(merged_params)
        
        return merged_model

    def save_state(self, path):
        """Save AWD state"""
        torch.save({
            'redundant_vector': self.redundant_vector,
            'disentangled_vectors': self.disentangled_vectors,
            'model_state': self.model.state_dict(),
        }, path)

    def load_state(self, path):
        """Load AWD state"""
        checkpoint = torch.load(path)
        self.redundant_vector = checkpoint['redundant_vector']
        self.disentangled_vectors = checkpoint['disentangled_vectors']
        self.model.load_state_dict(checkpoint['model_state'])
```
